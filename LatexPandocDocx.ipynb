```python

# ==============================================================================
# Google Colab Script: Convert LaTeX to Word (.docx) & Save .tex File
# ==============================================================================

import os
import ipywidgets as widgets
from IPython.display import display
from google.colab import files

print("📦 Step 1: Installing Pandoc and system dependencies...")
!apt-get update -qq > /dev/null
!apt-get install -y pandoc pandoc-citeproc > /dev/null
print("✅ Pandoc installed successfully!\n")

def process_tex_and_convert(tex_content, tex_filename="my_document.tex"):
    # Ensure proper file extensions
    if not tex_filename.endswith('.tex'):
        tex_filename += '.tex'
    docx_filename = tex_filename.replace('.tex', '.docx')

    # 💾 Step A: Save the LaTeX code to a local .tex file on Colab
    with open(tex_filename, "w", encoding="utf-8") as f:
        f.write(tex_content)
    print(f"\n💾 Saved LaTeX file to Colab workspace: '{tex_filename}'")

    # ⚙️ Step B: Convert .tex to .docx using Pandoc
    print(f"⚙️ Converting '{tex_filename}' to '{docx_filename}' using Pandoc...")
    cmd = f'pandoc "{tex_filename}" -s -o "{docx_filename}"'
    exit_code = os.system(cmd)

    # 📥 Step C: Download the generated .docx file
    if exit_code == 0 and os.path.exists(docx_filename):
        print(f"✅ Conversion successful! Downloading '{docx_filename}'...")
        files.download(docx_filename)
    else:
        print("❌ Conversion failed. Please check your LaTeX syntax for errors.")

# Interactive Mode Selection
print("==================================================")
mode = input("How would you like to provide your LaTeX document?\n [1] Upload a .tex file\n [2] Paste LaTeX code directly\n\nEnter 1 or 2: ").strip()

if mode == "1":
    print("\n📂 Step 2: Please upload your .tex file below...")
    uploaded = files.upload()
    for filename, content in uploaded.items():
        if filename.endswith('.tex'):
            tex_str = content.decode('utf-8')
            process_tex_and_convert(tex_str, filename)
        else:
            print(f"⚠️ Skipping '{filename}' (not a .tex file).")

elif mode == "2":
    print("\n📋 Step 2: Configure filename and paste your LaTeX code below:\n")

    filename_input = widgets.Text(
        value='my_document.tex',
        description='File Name:',
        placeholder='e.g. paper.tex',
        layout=widgets.Layout(width='50%')
    )

    text_area = widgets.Textarea(
        placeholder='Paste your complete LaTeX code here (e.g. \\documentclass{journal}...)...',
        layout=widgets.Layout(width='100%', height='380px')
    )

    convert_button = widgets.Button(
        description="💾 Save .tex & ⚙️ Convert to .docx",
        button_style='success',
        layout=widgets.Layout(width='280px', height='42px')
    )

    def on_button_click(b):
        code = text_area.value.strip()
        fname = filename_input.value.strip() or "my_document.tex"
        if not code:
            print("\n⚠️ The text area is empty! Please paste your LaTeX code first.")
        else:
            process_tex_and_convert(code, fname)

    convert_button.on_click(on_button_click)
    display(filename_input, text_area, convert_button)

else:
    print("\n❌ Invalid choice! Please re-run the cell and enter '1' or '2'.")
```

```python
# ==============================================================================
# Google Colab Script: Convert LaTeX to Word (.docx)
# Supports:
#   Option 1: Uploading a .tex file
#   Option 2: Pasting LaTeX code directly into a text area
# ==============================================================================

import os
import ipywidgets as widgets
from IPython.display import display
from google.colab import files

print("📦 Step 1: Installing Pandoc and system dependencies...")
!apt-get update -qq > /dev/null
!apt-get install -y pandoc pandoc-citeproc > /dev/null
print("✅ Pandoc installed successfully!\n")

# Helper function to convert LaTeX string to DOCX and download
def convert_and_download(tex_content, output_docx_name):
    temp_tex = "input_document.tex"
    with open(temp_tex, "w", encoding="utf-8") as f:
        f.write(tex_content)

    print(f"\n⚙️ Converting LaTeX to '{output_docx_name}' using Pandoc...")
    cmd = f'pandoc "{temp_tex}" -s -o "{output_docx_name}"'
    exit_code = os.system(cmd)

    if exit_code == 0 and os.path.exists(output_docx_name):
        print(f"✅ Conversion successful! Downloading '{output_docx_name}'...")
        files.download(output_docx_name)
    else:
        print("❌ Conversion failed. Please check your LaTeX code for syntax errors.")

# Interactive choice selection
print("==================================================")
mode = input("How would you like to provide your LaTeX document?\n [1] Upload a .tex file\n [2] Paste LaTeX code directly\n\nEnter 1 or 2: ").strip()

if mode == "1":
    print("\n📂 Step 2: Please upload your .tex file below...")
    uploaded = files.upload()
    for filename, content in uploaded.items():
        if filename.endswith('.tex'):
            tex_str = content.decode('utf-8')
            docx_name = filename.replace('.tex', '.docx')
            convert_and_download(tex_str, docx_name)
        else:
            print(f"⚠️ Skipping '{filename}' (not a .tex file).")

elif mode == "2":
    print("\n📋 Step 2: Paste your LaTeX code into the text area below and click 'Convert & Download':\n")

    text_area = widgets.Textarea(
        placeholder='Paste your complete LaTeX code here (e.g. \\documentclass{journal}... \\begin{document}...)...',
        layout=widgets.Layout(width='100%', height='380px')
    )

    convert_button = widgets.Button(
        description="⚙️ Convert & Download .docx",
        button_style='success',
        layout=widgets.Layout(width='250px', height='42px')
    )

    def on_button_click(b):
        code = text_area.value.strip()
        if not code:
            print("\n⚠️ The text area is empty! Please paste your LaTeX code first.")
        else:
            convert_and_download(code, "pasted_document.docx")

    convert_button.on_click(on_button_click)
    display(text_area, convert_button)

else:
    print("\n❌ Invalid choice! Please re-run the cell and enter '1' or '2'.")
```
